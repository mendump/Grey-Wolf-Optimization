```python
import os
import random
import numpy as np
import pandas as pd

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.ensemble import RandomForestRegressor, HistGradientBoostingRegressor
from sklearn.model_selection import train_test_split

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader

SEED = 42
CSV_PATH = "/content/merged_crop_yield_5000_unique.csv"
TARGET = "yield_t_ha"
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
def metrics(y_true, y_pred, label):
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    mae = mean_absolute_error(y_true, y_pred)
    r2 = r2_score(y_true, y_pred)
    print(f"{label:18s} MAE={mae:.4f} | RMSE={rmse:.4f} | R2={r2:.4f}")
    return {"model": label, "MAE": mae, "RMSE": rmse, "R2": r2}
```

```python
# 1. Load and validate data
df = pd.read_csv(CSV_PATH)
df.columns = [c.strip() for c in df.columns]

if TARGET not in df.columns:
    if {"production_t", "area_ha"}.issubset(df.columns):
        df[TARGET] = df["production_t"] / df["area_ha"].replace(0, np.nan)
    else:
        raise ValueError("CSV needs yield_t_ha or both production_t and area_ha.")

df = df.replace([np.inf, -np.inf], np.nan)
df = df.dropna(subset=[TARGET, "year"]).copy()
df = df[df[TARGET] >= 0].copy()
```

```python
# Do not let target leakage enter the predictors.
excluded = {
    TARGET, "production_t", "yield", "yield_tons_per_ha",
    "yield_kg_ha", "yield_t_ha"
}
# Area is retained as a predictor; production is excluded.
feature_cols = [c for c in df.columns if c not in excluded]
if "year" not in feature_cols:
    raise ValueError("A year column is required for temporal splitting.")
```

```python
# Use later years as a final temporal test set.
years = sorted(df["year"].dropna().unique())
if len(years) < 3:
    raise ValueError("Need at least 3 distinct years for train/validation/test.")
test_start = years[max(1, int(len(years) * 0.8))]
val_start = years[max(1, int(len(years) * 0.65))]

train_df = df[df["year"] < val_start].copy()
val_df = df[(df["year"] >= val_start) & (df["year"] < test_start)].copy()
test_df = df[df["year"] >= test_start].copy()

if min(len(train_df), len(val_df), len(test_df)) == 0:
    # Fallback chronological row split if year coverage is sparse.
    df = df.sort_values("year").reset_index(drop=True)
    n = len(df)
    train_df, val_df, test_df = (
        df.iloc[:int(.65*n)].copy(),
        df.iloc[int(.65*n):int(.80*n)].copy(),
        df.iloc[int(.80*n):].copy()
    )

X_train_raw = train_df[feature_cols]
X_val_raw = val_df[feature_cols]
X_test_raw = test_df[feature_cols]
y_train = train_df[TARGET].astype("float32").to_numpy()
y_val = val_df[TARGET].astype("float32").to_numpy()
y_test = test_df[TARGET].astype("float32").to_numpy()

cat_cols = X_train_raw.select_dtypes(include=["object", "category", "bool"]).columns.tolist()
num_cols = [c for c in feature_cols if c not in cat_cols]

numeric_pipe = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])
categorical_pipe = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore", sparse_output=False))
])
preprocessor = ColumnTransformer([
    ("num", numeric_pipe, num_cols),
    ("cat", categorical_pipe, cat_cols)
])

X_train = preprocessor.fit_transform(X_train_raw).astype("float32")
X_val = preprocessor.transform(X_val_raw).astype("float32")
X_test = preprocessor.transform(X_test_raw).astype("float32")
```

```python
# 2. Conventional baselines
results = []

rf = RandomForestRegressor(
    n_estimators=400, max_depth=None, min_samples_leaf=2,
    random_state=SEED, n_jobs=-1
)
rf.fit(X_train, y_train)
results.append(metrics(y_test, rf.predict(X_test), "Random Forest"))

hgb = HistGradientBoostingRegressor(
    max_iter=300, learning_rate=0.05, max_leaf_nodes=20,
    l2_regularization=1.0, random_state=SEED
)
hgb.fit(X_train, y_train)
results.append(metrics(y_test, hgb.predict(X_test), "HistGradientBoosting"))
```

```python
# 3. Grey Wolf Optimizer for feature selection
# Fitness is validation RMSE from a lightweight RF. Test data is never used.
def gwo_feature_selection(Xtr, ytr, Xv, yv, wolves=8, iterations=12):
    dim = Xtr.shape[1]
    rng = np.random.default_rng(SEED)
    positions = rng.uniform(0, 1, size=(wolves, dim))

    def evaluate(pos):
        mask = pos > 0.5
        if mask.sum() == 0:
            mask[np.argmax(pos)] = True
        model = RandomForestRegressor(
            n_estimators=100, min_samples_leaf=2,
            random_state=SEED, n_jobs=-1
        )
        model.fit(Xtr[:, mask], ytr)
        pred = model.predict(Xv[:, mask])
        # Small penalty discourages selecting every feature.
        return np.sqrt(mean_squared_error(yv, pred)) + 0.002 * mask.sum()

    fitness = np.array([evaluate(p) for p in positions])
    order = np.argsort(fitness)
    alpha, beta, delta = positions[order[0]].copy(), positions[order[1]].copy(), positions[order[2]].copy()
    alpha_score = fitness[order[0]]

    for t in range(iterations):
        a = 2 - 2 * (t / max(iterations - 1, 1))
        for i in range(wolves):
            candidates = []
            for leader in (alpha, beta, delta):
                r1, r2 = rng.random(dim), rng.random(dim)
                A = 2 * a * r1 - a
                C = 2 * r2
                D = np.abs(C * leader - positions[i])
                candidates.append(leader - A * D)
            positions[i] = np.clip(np.mean(candidates, axis=0), 0, 1)

        fitness = np.array([evaluate(p) for p in positions])
        order = np.argsort(fitness)
        alpha, beta, delta = (positions[order[j]].copy() for j in range(3))
        alpha_score = fitness[order[0]]
        print(f"GWO iteration {t+1:02d}/{iterations}: validation fitness={alpha_score:.4f}")

    mask = alpha > 0.5
    if mask.sum() == 0:
        mask[np.argmax(alpha)] = True
    return mask


selected = gwo_feature_selection(X_train, y_train, X_val, y_val)
print(f"GWO selected {selected.sum()} of {len(selected)} encoded features.")
```

```python
# 4. Feature-token Transformer regressor
class TabularDataset(Dataset):
    def __init__(self, X, y=None):
        self.X = torch.tensor(X, dtype=torch.float32)
        self.y = None if y is None else torch.tensor(y, dtype=torch.float32)

    def __len__(self):
        return len(self.X)

    def __getitem__(self, idx):
        if self.y is None:
            return self.X[idx]
        return self.X[idx], self.y[idx]


class FeatureTokenTransformer(nn.Module):
    def __init__(self, n_features, d_model=64, heads=4, layers=2, dropout=0.15):
        super().__init__()
        self.value_embed = nn.Linear(1, d_model)
        self.feature_embed = nn.Parameter(torch.randn(1, n_features, d_model) * 0.02)
        enc = nn.TransformerEncoderLayer(
            d_model=d_model, nhead=heads, dim_feedforward=128,
            dropout=dropout, batch_first=True, activation="gelu"
        )
        self.encoder = nn.TransformerEncoder(enc, num_layers=layers)
        self.norm = nn.LayerNorm(d_model)
        self.head = nn.Sequential(
            nn.Linear(d_model, 64), nn.GELU(),
            nn.Dropout(dropout), nn.Linear(64, 1)
        )

    def forward(self, x):
        # Each standardized tabular feature becomes one token.
        tokens = self.value_embed(x.unsqueeze(-1)) + self.feature_embed
        z = self.encoder(tokens)
        z = self.norm(z).mean(dim=1)
        return self.head(z).squeeze(-1)


def train_transformer(Xtr, ytr, Xv, yv, epochs=80, lr=1e-3):
    model = FeatureTokenTransformer(Xtr.shape[1]).to(DEVICE)
    train_loader = DataLoader(TabularDataset(Xtr, ytr), batch_size=32, shuffle=True)
    val_loader = DataLoader(TabularDataset(Xv, yv), batch_size=128, shuffle=False)
    optimizer = torch.optim.AdamW(model.parameters(), lr=lr, weight_decay=1e-4)
    loss_fn = nn.MSELoss()
    best_state, best_val = None, float("inf")
    patience, wait = 12, 0

    for epoch in range(epochs):
        model.train()
        for xb, yb in train_loader:
            xb, yb = xb.to(DEVICE), yb.to(DEVICE)
            optimizer.zero_grad()
            loss = loss_fn(model(xb), yb)
            loss.backward()
            optimizer.step()

        model.eval()
        pred_all, actual_all = [], []
        with torch.no_grad():
            for xb, yb in val_loader:
                pred_all.extend(model(xb.to(DEVICE)).cpu().numpy())
                actual_all.extend(yb.numpy())
        val_rmse = np.sqrt(mean_squared_error(actual_all, pred_all))
        if val_rmse < best_val:
            best_val, wait = val_rmse, 0
            best_state = {k: v.detach().cpu().clone() for k, v in model.state_dict().items()}
        else:
            wait += 1
        if wait >= patience:
            break

    model.load_state_dict(best_state)
    return model


Xtr_gwo, Xv_gwo, Xt_gwo = X_train[:, selected], X_val[:, selected], X_test[:, selected]
vit_gwo = train_transformer(Xtr_gwo, y_train, Xv_gwo, y_val)
test_loader = DataLoader(TabularDataset(Xt_gwo), batch_size=128, shuffle=False)
vit_gwo.eval()
predictions = []
with torch.no_grad():
    for xb in test_loader:
        predictions.extend(vit_gwo(xb.to(DEVICE)).cpu().numpy())

results.append(metrics(y_test, np.asarray(predictions), "ViT-GWO"))

pd.DataFrame(results).to_csv("model_comparison_results.csv", index=False)
pd.DataFrame({
    "actual_yield_t_ha": y_test,
    "predicted_yield_t_ha": predictions
}).to_csv("vit_gwo_test_predictions.csv", index=False)

print("\nSaved: model_comparison_results.csv")
print("Saved: vit_gwo_test_predictions.csv")
print("Device:", DEVICE)
```

```python
import os, warnings
warnings.filterwarnings("ignore")
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.ensemble import RandomForestRegressor, HistGradientBoostingRegressor
from sklearn.model_selection import TimeSeriesSplit, cross_val_score

SEED = 42
CSV = "/merged_crop_yield_5000_unique.csv"
OUT = "crop_yield_results"
TARGET = "yield_t_ha"
os.makedirs(OUT, exist_ok=True)
df = pd.read_csv(CSV)

cols = ["year","district","crop","rainfall_mm","tmean_c","soil_ph","NDVI_mean","yield_t_ha"]
missing = [c for c in cols if c not in df.columns]
if missing:
    raise ValueError("Missing columns: " + str(missing))
df = df[cols].copy()
df["year"] = pd.to_numeric(df["year"], errors="coerce")
for c in ["rainfall_mm","tmean_c","soil_ph","NDVI_mean",TARGET]:
    df[c] = pd.to_numeric(df[c], errors="coerce")
df["district"] = df["district"].astype("string").str.strip()
df["crop"] = df["crop"].astype("string").str.strip()
df = df.dropna(subset=["year","district",TARGET])
df = df[df.crop.str.lower() == "rice"].sort_values(["year","district"]).reset_index(drop=True)
if len(df) < 30 or df.year.nunique() < 3:
    raise ValueError("Need at least 30 usable rows and 3 different years.")
```

```python
# Hold out the latest 20% of years to avoid random temporal leakage.
years = np.sort(df.year.unique())
k = max(1, int(len(years)*0.8))
test_years = years[k:] if k < len(years) else years[-1:]
tr, te = df[~df.year.isin(test_years)].copy(), df[df.year.isin(test_years)].copy()
features = ["year","district","rainfall_mm","tmean_c","soil_ph","NDVI_mean"]
num = ["year","rainfall_mm","tmean_c","soil_ph","NDVI_mean"]
try:
    encoder = OneHotEncoder(handle_unknown="ignore", sparse_output=False)
except TypeError:
    encoder = OneHotEncoder(handle_unknown="ignore", sparse=False)
prep = ColumnTransformer([
    ("num", Pipeline([("impute",SimpleImputer(strategy="median")),
                      ("scale",StandardScaler())]), num),
    ("district", Pipeline([("impute",SimpleImputer(strategy="most_frequent")),
                           ("onehot",encoder)]), ["district"])
])
Xtr = prep.fit_transform(tr[features]).astype("float32")
Xte = prep.transform(te[features]).astype("float32")
ytr, yte = tr[TARGET].to_numpy("float32"), te[TARGET].to_numpy("float32")

def metrics(name, actual, pred):
    return {"Method":name,
            "MAE":mean_absolute_error(actual,pred),
            "RMSE":mean_squared_error(actual,pred)**0.5,
            "R2":r2_score(actual,pred)}

preds, rows = {}, []
```

```python
# Existing methodology
baselines = {
    "Random Forest (Existing)": RandomForestRegressor(
        n_estimators=250,min_samples_leaf=2,max_features=0.9,
        random_state=SEED,n_jobs=-1),
    "HistGradientBoosting (Existing)": HistGradientBoostingRegressor(
        max_iter=200,learning_rate=0.05,max_leaf_nodes=15,
        l2_regularization=1.0,random_state=SEED)
}
for name, model in baselines.items():
    model.fit(Xtr,ytr)
    preds[name] = model.predict(Xte)
    rows.append(metrics(name,yte,preds[name]))
```

```python
# Grey Wolf Optimizer feature selection, with time-series CV fitness.
def gwo_select(X,y,wolves=6,iterations=8):
    n=X.shape[1]
    rng=np.random.default_rng(SEED)
    pos=rng.random((wolves,n))
    cv=TimeSeriesSplit(n_splits=3)
    def fitness(v):
        mask=v>0.5
        if not mask.any(): mask[np.argmax(v)]=True
        est=HistGradientBoostingRegressor(max_iter=80,learning_rate=0.06,
                                          max_leaf_nodes=10,random_state=SEED)
        score=cross_val_score(est,X[:,mask],y,cv=cv,
             scoring="neg_root_mean_squared_error",n_jobs=1)
        return -score.mean()+0.01*mask.mean()
    best=pos[0].copy(); bestfit=float("inf"); history=[]
    for it in range(iterations):
        fs=np.array([fitness(v) for v in pos])
        order=np.argsort(fs)
        alpha,beta,delta=[pos[j].copy() for j in order[:3]]
        if fs[order[0]]<bestfit:
            bestfit=float(fs[order[0]]); best=alpha.copy()
        a=2-2*it/max(1,iterations-1)
        new=pos.copy()
        for i in range(wolves):
            candidates=[]
            for leader in (alpha,beta,delta):
                r1=rng.random(n); r2=rng.random(n)
                A=2*a*r1-a; C=2*r2
                candidates.append(leader-A*np.abs(C*leader-pos[i]))
            new[i]=np.clip(np.mean(candidates,axis=0),0,1)
        pos=new; history.append(bestfit)
    mask=best>0.5
    if not mask.any(): mask[np.argmax(best)]=True
    return mask,history

mask,gwo_curve=gwo_select(Xtr,ytr)
A,B=Xtr[:,mask],Xte[:,mask]
```

```python
# Proposed feature-token Transformer (tabular Transformer).
import torch
from torch import nn
from torch.utils.data import TensorDataset,DataLoader
torch.manual_seed(SEED)
device=torch.device("cuda" if torch.cuda.is_available() else "cpu")

class FeatureTransformer(nn.Module):
    def __init__(self,n):
        super().__init__()
        d=32
        self.proj=nn.Linear(1,d)
        self.feature_embed=nn.Parameter(torch.randn(1,n,d)*0.02)
        enc=nn.TransformerEncoderLayer(d_model=d,nhead=4,dim_feedforward=64,
             dropout=0.15,batch_first=True,activation="gelu")
        self.encoder=nn.TransformerEncoder(enc,num_layers=2)
        self.head=nn.Sequential(nn.LayerNorm(d),nn.Linear(d,32),nn.GELU(),
                                 nn.Dropout(0.15),nn.Linear(32,1))
    def forward(self,x):
        z=self.proj(x.unsqueeze(-1))+self.feature_embed
        return self.head(self.encoder(z).mean(dim=1)).squeeze(-1)

net=FeatureTransformer(A.shape[1]).to(device)
opt=torch.optim.AdamW(net.parameters(),lr=0.001,weight_decay=1e-4)
lossfn=nn.MSELoss()
nval=max(1,int(0.15*len(A)))
nfit=len(A)-nval
if nfit<5: nfit=len(A); nval=0
ds=TensorDataset(torch.tensor(A[:nfit]),torch.tensor(ytr[:nfit]))
loader=DataLoader(ds,batch_size=min(32,nfit),shuffle=True)
if nval:
    xv=torch.tensor(A[nfit:],dtype=torch.float32).to(device)
    yv=torch.tensor(ytr[nfit:],dtype=torch.float32).to(device)
trainloss=[]; valloss=[]; beststate=None; bestval=float("inf"); patience=0
for epoch in range(160):
    net.train(); losses=[]
    for xb,yb in loader:
        xb=xb.to(device); yb=yb.to(device)
        opt.zero_grad(); loss=lossfn(net(xb),yb)
        loss.backward(); torch.nn.utils.clip_grad_norm_(net.parameters(),1.0)
        opt.step(); losses.append(loss.item())
    trainloss.append(float(np.mean(losses)))
    if nval:
        net.eval()
        with torch.no_grad(): vl=lossfn(net(xv),yv).item()
        valloss.append(vl)
        if vl<bestval:
            bestval=vl
            beststate={k:v.detach().cpu().clone() for k,v in net.state_dict().items()}
            patience=0
        else:
            patience+=1
            if patience>=18: break
if beststate is not None: net.load_state_dict(beststate)
net.eval()
with torch.no_grad():
    proposed=net(torch.tensor(B,dtype=torch.float32).to(device)).cpu().numpy()
preds["Proposed: GWO + Transformer"]=proposed
rows.append(metrics("Proposed: GWO + Transformer",yte,proposed))
```

```python
# Save numerical results and predictions.
result=pd.DataFrame(rows).sort_values("RMSE")
result.to_csv(os.path.join(OUT,"model_comparison_metrics.csv"),index=False)
out=te[["year","district","crop",TARGET]].copy()
for name,p in preds.items():
    col=name.lower().replace(" ","_").replace(":","").replace("+","plus")
    out[col]=p
out.to_csv(os.path.join(OUT,"test_predictions.csv"),index=False)
```

```python
# Charts based on the actual held-out test predictions.
ix=np.arange(len(result)); w=0.25
plt.figure(figsize=(9,5))
plt.bar(ix-w/2,result.MAE,w,label="MAE")
plt.bar(ix+w/2,result.RMSE,w,label="RMSE")
plt.xticks(ix,result.Method,rotation=12,ha="right")
plt.ylabel("Error (tonnes/hectare)"); plt.title("Existing vs Proposed: MAE and RMSE")
plt.legend(); plt.tight_layout()
plt.savefig(os.path.join(OUT,"01_error_comparison.png"),dpi=300); plt.close()

plt.figure(figsize=(8,5))
plt.bar(result.Method,result.R2); plt.axhline(0,ls="--",lw=0.8)
plt.ylabel("R²"); plt.title("R² Comparison")
plt.xticks(rotation=12,ha="right"); plt.tight_layout()
plt.savefig(os.path.join(OUT,"02_r2_comparison.png"),dpi=300); plt.close()

plt.figure(figsize=(7,5))
plt.plot(range(1,len(gwo_curve)+1),gwo_curve,marker="o")
plt.xlabel("GWO iteration"); plt.ylabel("Best CV fitness (lower is better)")
plt.title("GWO Convergence"); plt.tight_layout()
plt.savefig(os.path.join(OUT,"03_gwo_convergence.png"),dpi=300); plt.close()

plt.figure(figsize=(7,5))
plt.plot(trainloss,label="Training loss")
if valloss: plt.plot(valloss,label="Validation loss")
plt.xlabel("Epoch"); plt.ylabel("MSE loss"); plt.title("Transformer Training Curve")
plt.legend(); plt.tight_layout()
plt.savefig(os.path.join(OUT,"04_training_curve.png"),dpi=300); plt.close()

plt.figure(figsize=(7,6))
for name,p in preds.items(): plt.scatter(yte,p,alpha=0.65,label=name)
lo=min(float(yte.min()),min(float(p.min()) for p in preds.values()))
hi=max(float(yte.max()),max(float(p.max()) for p in preds.values()))
plt.plot([lo,hi],[lo,hi],"k--",label="Ideal")
plt.xlabel("Actual yield (t/ha)"); plt.ylabel("Predicted yield (t/ha)")
plt.title("Actual vs Predicted Rice Yield"); plt.legend(fontsize=8)
plt.tight_layout(); plt.savefig(os.path.join(OUT,"05_actual_vs_predicted.png"),dpi=300)
plt.close()

print("\nCOMPUTED TEST RESULTS")
print(result.to_string(index=False))
print(f"Train rows={len(tr)}, test rows={len(te)}")
print(f"Train years={tr.year.min():.0f}-{tr.year.max():.0f}; test years={te.year.min():.0f}-{te.year.max():.0f}")
print(f"GWO selected {int(mask.sum())}/{len(mask)} encoded features.")
print("Charts and CSV results saved in:",OUT)
```
